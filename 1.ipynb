```python
import pandas as pd
merc=pd.read_csv("merc.csv")
merc.head()
```

```python
merc.isnull().sum()
# Menda bo'sh qiymatlar yo'q ekan
```

```python
merc.duplicated().sum()
```

```python
merc = merc.drop_duplicates()
merc.duplicated().sum()
```

```python
merc.info()
```

```python
merc.fillna(merc.mean(numeric_only=True), inplace=True)
# bo'sh qiymatlar yo'q lekin agar menda bo'sh qiymatlar bo'lganda uni to'lddirib berar edi
```

```python
merc.fillna(merc.median(numeric_only=True), inplace=True)
# bo'sh qiymatlar yo'q lekin agar menda bo'sh qiymatlar bo'lganda uni to'lddirib berar edi
```

```python
merc.fillna(merc.mode().iloc[0], inplace=True)
# bo'sh qiymatlar yo'q lekin agar menda bo'sh qiymatlar bo'lganda uni to'lddirib berar edi
```
